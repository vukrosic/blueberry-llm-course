# Tokenization

In this lesson, we will learn how to convert text into numbers that our model can understand. We will use the tokenizer from `SmolLM-135M`, which is the model we are replicating.

```python
from transformers import AutoTokenizer

# Load the tokenizer
tokenizer = AutoTokenizer.from_pretrained("HuggingFaceTB/SmolLM-135M")

print(f"Vocab size: {tokenizer.vocab_size}")
```

## Encoding Text

Let's see how the tokenizer breaks down a sentence into tokens.

```python
text = "Hello, world! This is a test."
tokens = tokenizer.encode(text)

print(f"Text: {text}")
print(f"Tokens: {tokens}")
```

## Decoding Tokens

We can also convert the tokens back into text.

```python
decoded_text = tokenizer.decode(tokens)
print(f"Decoded: {decoded_text}")
```

## Special Tokens

The tokenizer handles special tokens like padding and end-of-sequence.

```python
print(f"EOS Token: {tokenizer.eos_token} (ID: {tokenizer.eos_token_id})")
print(f"Pad Token: {tokenizer.pad_token} (ID: {tokenizer.pad_token_id})")
```
