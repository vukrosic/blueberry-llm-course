# RMSNorm (Root Mean Square Normalization)

RMSNorm is a normalization technique often used in modern LLMs (like LLaMA) instead of LayerNorm. It is computationally simpler and often performs just as well.

## Formula

$$ \text{RMSNorm}(x) = \frac{x}{\sqrt{\text{mean}(x^2) + \epsilon}} * \gamma $$

Where $\gamma$ is a learnable parameter.

```python
import torch
import torch.nn as nn

class RMSNorm(nn.Module):
    def __init__(self, dim, eps=1e-6):
        super().__init__()
        self.eps = eps
        self.weight = nn.Parameter(torch.ones(dim))

    def forward(self, x):
        # Calculate RMS
        rms = torch.sqrt(torch.mean(x**2, dim=-1, keepdim=True) + self.eps)
        # Normalize and scale
        return x / rms * self.weight

# Test it
x = torch.randn(2, 5)
norm = RMSNorm(5)
output = norm(x)
print("Input:", x)
print("Output:", output)
```

## PyTorch Implementation

PyTorch now has a built-in `nn.RMSNorm`, which is what we use in our model.

```python
torch_norm = nn.RMSNorm(5)
output_torch = torch_norm(x)
print("PyTorch Output:", output_torch)
```
