# Rotary Positional Embeddings (RoPE)

RoPE is a method for encoding position information in Transformers. Instead of adding a position vector to the embedding (like in the original Transformer), RoPE rotates the query and key vectors.

## How it works

RoPE applies a rotation matrix to pairs of elements in the vector. The angle of rotation depends on the position in the sequence.

```python
import torch
from torchtune.modules import RotaryPositionalEmbeddings

# Parameters
dim = 64
max_seq_len = 512

# Initialize RoPE
rope = RotaryPositionalEmbeddings(dim=dim, max_seq_len=max_seq_len)

# Create dummy Query and Key
batch_size, seq_len, num_heads, head_dim = 1, 10, 4, 64
q = torch.randn(batch_size, seq_len, num_heads, head_dim)
k = torch.randn(batch_size, seq_len, num_heads, head_dim)

# Apply RoPE
q_rotated = rope(q)
k_rotated = rope(k)

print("Original shape:", q.shape)
print("Rotated shape:", q_rotated.shape)
```

## Why RoPE?

RoPE allows the model to understand relative positions better than absolute positions. If token A is at position 5 and token B is at position 10, the relationship is preserved if we shift them to 105 and 110.
