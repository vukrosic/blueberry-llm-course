# SiLU (Sigmoid Linear Unit)

SiLU (also known as Swish) is an activation function used in many state-of-the-art models. It is defined as:

$$ \text{SiLU}(x) = x * \sigma(x) $$

where $\sigma(x)$ is the sigmoid function.

```python
import torch
import torch.nn.functional as F
import matplotlib.pyplot as plt

x = torch.linspace(-5, 5, 100)
y = F.silu(x)

plt.plot(x.numpy(), y.numpy())
plt.title("SiLU Activation")
plt.grid(True)
plt.show()
```

## Why SiLU?

Unlike ReLU, which is zero for all negative inputs, SiLU allows a small amount of negative information to flow through, which can help with gradient flow during training. It is smooth and non-monotonic.
